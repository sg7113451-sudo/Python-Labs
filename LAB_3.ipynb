```python
train_name= "rajdhani express"
seat_number = "B2-45"
ticket_price = 2450
Journey_date = "25-May-2026"
```

```python
print("train_name:", train_name)
print("seat_number:", seat_number)
print("ticket_price:", ticket_price)
print("Journey_date:", Journey_date)
```

```python
number_of_passengers = 4
total_fare =  number_of_passengers * ticket_price
print("total fare       :", total_fare)
gst_percentage = 5
gst_amount = (total_fare * gst_percentage) / 100
print("gst_amount         :", gst_amount)

```

```python
final_amount = total_fare + gst_amount
print("final_payable_amount         :", final_amount)
```
